```python
# ==========================================
# STEP 3: DEFINE DATASET FILE PATH
# ==========================================

# Define the raw dataset folder path
RAW_DATA_PATH = os.path.join(
    PROJECT_PATH,
    "Dataset",
    "Raw"
)

# Display the files available in the raw dataset folder
print("Dataset Folder:", RAW_DATA_PATH)
print("\nAvailable Dataset Files:")

print(os.listdir(RAW_DATA_PATH))
```

```python
# ==========================================
# STEP 4: LOAD RETAIL SALES DATASET
# ==========================================

# Define the CSV file path
DATASET_FILE = os.path.join(
    RAW_DATA_PATH,
    "Sample - Superstore.csv"
)

# Load the CSV file into a Pandas DataFrame
df = pd.read_csv(DATASET_FILE, encoding="latin1")

# Display a success message
print("Retail dataset loaded successfully!")

# Display the number of rows and columns
print("\nDataset Shape:", df.shape)

# Display the first five records
print("\nFirst Five Records:")
display(df.head())
```

```python
# ==========================================
# STEP 5: DATASET STRUCTURE ANALYSIS
# ==========================================

# Display dataset information
print("DATASET INFORMATION")
print("-" * 40)

# Check column names, data types and non-null values
df.info()

# Display total rows and columns
print("\nDataset Shape:", df.shape)

# Display all column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display statistical summary of numerical columns
print("\nStatistical Summary:")
display(df.describe())
```

```python
# ==========================================
# STEP 6: MISSING VALUES ANALYSIS
# ==========================================

# Count missing values in each column
missing_values = df.isnull().sum()

# Calculate missing value percentage
missing_percentage = (missing_values / len(df)) * 100

# Create a missing value summary
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage.round(2)
})

# Display only columns containing missing values
missing_summary = missing_summary[
    missing_summary["Missing Values"] > 0
]

# Display the missing value report
print("MISSING VALUES REPORT")
display(missing_summary)
```

```python
# ==========================================
# STEP 7: DUPLICATE RECORDS ANALYSIS
# ==========================================

# Count the total number of duplicate rows
duplicate_count = df.duplicated().sum()

# Calculate the percentage of duplicate rows
duplicate_percentage = (duplicate_count / len(df)) * 100

# Display duplicate record analysis
print("DUPLICATE RECORDS REPORT")
print("-" * 40)
print("Total Records:", len(df))
print("Duplicate Records:", duplicate_count)
print(f"Duplicate Percentage: {duplicate_percentage:.2f}%")
```

```python
# ==========================================
# STEP 8: UNIQUE VALUE ANALYSIS
# ==========================================

# Select important categorical columns
categorical_columns = [
    "Category",
    "Sub-Category",
    "Region",
    "Segment",
    "Ship Mode"
]

# Count unique values in each selected column
for column in categorical_columns:

    print(f"\n{column}: {df[column].nunique()} Unique Values")

    # Display available categories
    print(df[column].unique())
```

```python
# ==========================================
# STEP 9: SALES AND PROFIT KPI ANALYSIS
# ==========================================

# Calculate total sales
total_sales = df["Sales"].sum()

# Calculate total profit
total_profit = df["Profit"].sum()

# Calculate total orders
total_orders = df["Order ID"].nunique()

# Calculate total quantity sold
total_quantity = df["Quantity"].sum()

# Calculate overall profit margin
profit_margin = (total_profit / total_sales) * 100

# Display business performance KPIs
print("RETAIL BUSINESS PERFORMANCE")
print("-" * 40)
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Quantity Sold: {total_quantity:,}")
print(f"Profit Margin: {profit_margin:.2f}%")
```

```python
# ==========================================
# STEP 10: MONTHLY SALES TREND ANALYSIS
# ==========================================

# Convert Order Date into datetime format
df["Order Date"] = pd.to_datetime(df["Order Date"])

# Extract month and year from Order Date
df["Order Month"] = df["Order Date"].dt.to_period("M")

# Calculate total sales for each month
monthly_sales = (
    df.groupby("Order Month")["Sales"]
    .sum()
    .reset_index()
)

# Rename columns for reporting
monthly_sales.columns = ["Month", "Total Sales"]

# Convert month values into strings
monthly_sales["Month"] = monthly_sales["Month"].astype(str)

# Display monthly sales report
display(monthly_sales)
```

```python
# ==========================================
# STEP 11: MONTHLY SALES TREND VISUALIZATION
# ==========================================

# Create a figure for the monthly sales chart
plt.figure(figsize=(14, 6))

# Plot monthly sales using a line chart
sns.lineplot(
    data=monthly_sales,
    x="Month",
    y="Total Sales",
    marker="o",
    linewidth=2
)

# Add chart title and axis labels
plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Month", fontsize=12)
plt.ylabel("Total Sales ($)", fontsize=12)

# Rotate month labels for readability
plt.xticks(rotation=45)

# Add grid lines
plt.grid(True, alpha=0.3)

# Adjust chart layout
plt.tight_layout()

# Display the chart
plt.show()
```

```python
# ==========================================
# STEP 12: CATEGORY-WISE SALES AND PROFIT ANALYSIS
# ==========================================

category_analysis = (
    df.groupby("Category")[["Sales", "Profit"]]
    .sum()
    .sort_values(by="Sales", ascending=False)
)

print("CATEGORY-WISE BUSINESS PERFORMANCE")
display(category_analysis)
```

```python
# ==========================================
# STEP 13: CATEGORY-WISE SALES AND PROFIT CHART
# ==========================================

# Create two side-by-side charts
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Sales by category
sns.barplot(
    data=category_analysis.reset_index(),
    x="Category",
    y="Sales",
    ax=axes[0]
)
axes[0].set_title("Sales by Product Category")
axes[0].set_xlabel("Category")
axes[0].set_ylabel("Total Sales ($)")

# Profit by category
sns.barplot(
    data=category_analysis.reset_index(),
    x="Category",
    y="Profit",
    ax=axes[1]
)
axes[1].set_title("Profit by Product Category")
axes[1].set_xlabel("Category")
axes[1].set_ylabel("Total Profit ($)")

# Adjust layout and display charts
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 14: SUB-CATEGORY SALES AND PROFIT ANALYSIS
# ==========================================

subcategory_analysis = (
    df.groupby("Sub-Category")[["Sales", "Profit"]]
    .sum()
    .sort_values(by="Profit", ascending=False)
)

print("SUB-CATEGORY-WISE BUSINESS PERFORMANCE")
display(subcategory_analysis)
```

```python
# ==========================================
# STEP 15: SUB-CATEGORY SALES AND PROFIT CHART
# ==========================================

subcategory_chart = subcategory_analysis.reset_index()

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Sales by sub-category
sales_data = subcategory_chart.sort_values("Sales", ascending=False)

sns.barplot(
    data=sales_data,
    x="Sales",
    y="Sub-Category",
    ax=axes[0]
)
axes[0].set_title("Sales by Sub-Category")
axes[0].set_xlabel("Total Sales ($)")
axes[0].set_ylabel("Sub-Category")

# Profit by sub-category
profit_data = subcategory_chart.sort_values("Profit", ascending=False)

sns.barplot(
    data=profit_data,
    x="Profit",
    y="Sub-Category",
    ax=axes[1]
)
axes[1].set_title("Profit by Sub-Category")
axes[1].set_xlabel("Total Profit ($)")
axes[1].set_ylabel("Sub-Category")

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 16: REGION-WISE SALES AND PROFIT ANALYSIS
# ==========================================

region_analysis = (
    df.groupby("Region")[["Sales", "Profit"]]
    .sum()
    .sort_values(by="Sales", ascending=False)
)

print("REGION-WISE BUSINESS PERFORMANCE")
display(region_analysis)
```

```python
# ==========================================
# STEP 17: REGION-WISE SALES AND PROFIT CHART
# ==========================================

region_chart = region_analysis.reset_index()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Sales by region
sns.barplot(
    data=region_chart,
    x="Region",
    y="Sales",
    ax=axes[0]
)
axes[0].set_title("Sales by Region")
axes[0].set_xlabel("Region")
axes[0].set_ylabel("Total Sales ($)")

# Profit by region
sns.barplot(
    data=region_chart,
    x="Region",
    y="Profit",
    ax=axes[1]
)
axes[1].set_title("Profit by Region")
axes[1].set_xlabel("Region")
axes[1].set_ylabel("Total Profit ($)")

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 18: DISCOUNT IMPACT ON PROFIT ANALYSIS
# ==========================================

discount_analysis = (
    df.groupby("Discount")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Orders=("Order ID", "nunique")
    )
    .reset_index()
    .sort_values("Discount")
)

print("DISCOUNT IMPACT ON BUSINESS PERFORMANCE")
display(discount_analysis)
```

```python
# ==========================================
# STEP 19: DISCOUNT IMPACT ON PROFIT CHART
# ==========================================

plt.figure(figsize=(10, 5))

sns.barplot(
    data=discount_analysis,
    x="Discount",
    y="Total_Profit"
)

plt.title("Impact of Discount on Total Profit")
plt.xlabel("Discount Level")
plt.ylabel("Total Profit ($)")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 20: TOP 10 PRODUCTS BY SALES
# ==========================================

top_products = (
    df.groupby("Product Name")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .sort_values(by="Total_Sales", ascending=False)
    .head(10)
    .reset_index()
)

print("TOP 10 PRODUCTS BY SALES")
display(top_products)
```

```python
# ==========================================
# STEP 21: TOP 10 PRODUCTS SALES CHART
# ==========================================

top_products_chart = top_products.sort_values(
    by="Total_Sales",
    ascending=True
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_products_chart,
    x="Total_Sales",
    y="Product Name"
)

plt.title("Top 10 Products by Sales", fontsize=16)
plt.xlabel("Total Sales ($)")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 22: CUSTOMER SEGMENT ANALYSIS
# ==========================================

segment_analysis = (
    df.groupby("Segment")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Orders=("Order ID", "nunique"),
        Total_Quantity=("Quantity", "sum")
    )
    .sort_values(by="Total_Sales", ascending=False)
    .reset_index()
)

print("CUSTOMER SEGMENT PERFORMANCE")
display(segment_analysis)
```

```python
# ==========================================
# STEP 23: CUSTOMER SEGMENT PERFORMANCE CHART
# ==========================================

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Sales by customer segment
sns.barplot(
    data=segment_analysis,
    x="Segment",
    y="Total_Sales",
    ax=axes[0]
)
axes[0].set_title("Sales by Customer Segment")
axes[0].set_xlabel("Customer Segment")
axes[0].set_ylabel("Total Sales ($)")

# Profit by customer segment
sns.barplot(
    data=segment_analysis,
    x="Segment",
    y="Total_Profit",
    ax=axes[1]
)
axes[1].set_title("Profit by Customer Segment")
axes[1].set_xlabel("Customer Segment")
axes[1].set_ylabel("Total Profit ($)")

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 24: SHIPPING MODE PERFORMANCE ANALYSIS
# ==========================================

shipping_analysis = (
    df.groupby("Ship Mode")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Orders=("Order ID", "nunique")
    )
    .sort_values(by="Total_Sales", ascending=False)
    .reset_index()
)

print("SHIPPING MODE PERFORMANCE")
display(shipping_analysis)
```

```python
# ==========================================
# STEP 25: SHIPPING MODE PERFORMANCE CHART
# ==========================================

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Sales by shipping mode
sns.barplot(
    data=shipping_analysis,
    x="Ship Mode",
    y="Total_Sales",
    ax=axes[0]
)
axes[0].set_title("Sales by Shipping Mode")
axes[0].set_xlabel("Shipping Mode")
axes[0].set_ylabel("Total Sales ($)")
axes[0].tick_params(axis="x", rotation=20)

# Profit by shipping mode
sns.barplot(
    data=shipping_analysis,
    x="Ship Mode",
    y="Total_Profit",
    ax=axes[1]
)
axes[1].set_title("Profit by Shipping Mode")
axes[1].set_xlabel("Shipping Mode")
axes[1].set_ylabel("Total Profit ($)")
axes[1].tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.show()
```
